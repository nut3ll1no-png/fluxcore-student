```python
# FluxCore - Primo modello AI + Energy Simulator
!pip install -q transformers

from transformers import pipeline

# Carichiamo un modello piccolo e veloce (distilgpt2)
generator = pipeline('text-generation', model='distilgpt2')

print("=== FluxCore AI Energy Assistant ===\n")
prompt = "Suggerisci 3 modi per ridurre il consumo energetico di un cluster GPU con 8 schede H100 al 30% di utilizzo:"
result = generator(prompt, max_length=150, num_return_sequences=1)

print(result[0]['generated_text'])
```

```python
# Collega l'AI al tuo simulatore energetico
def calcola_consumo():
    num_gpus = 8
    watt_per_gpu = 699
    utilizzo = 0.30
    sprecata = num_gpus * watt_per_gpu * (1 - utilizzo) / 1000
    print(f"Energia sprecata: {sprecata:.2f} kWh/ora")

calcola_consumo()
print("\nAI sta analizzando come ottimizzare...")
```
